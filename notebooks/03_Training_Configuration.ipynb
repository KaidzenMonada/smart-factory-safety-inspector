# Training Configuration

This notebook documents the final model-comparison setup for the Smart Factory Safety Inspector project. It contains no executable code.

## Objective

Compare six YOLO candidate models on the same cleaned PPE dataset using a shared training configuration. The goal is to select the best model for person-centered PPE safety inspection while keeping the comparison controlled and reproducible.

## Final Dataset

| Split | Images |
|---|---:|
| train | 3,219 |
| valid | 323 |
| test | 323 |

Dataset YAML: `data/final_dataset/data.yaml`

Classes: `person`, `helmet`, `vest`, `gloves`, `goggles`.

Every candidate model receives the same dataset, train/validation split, image size, training length, batch size, patience, augmentation settings, seed, AMP setting, and deterministic setting.

## Comparison Fairness Principle

The only planned difference between runs is the candidate model checkpoint. All training arguments are shared so the comparison reflects model-family and model-size behavior rather than differences in dataset preparation or hyperparameter tuning.

Training is not started in this stage.
