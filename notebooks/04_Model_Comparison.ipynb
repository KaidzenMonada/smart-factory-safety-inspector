# Model Comparison and Selection

This notebook summarizes Stage 4 model comparison for the Smart Factory Safety Inspector project. It contains no executable code and does not use the test split.

## Objective

Compare the six trained candidate checkpoints on validation performance, model complexity, and practical deployment efficiency, then select one final model for Stage 5 final evaluation.

## Per-Class Comparison

Per-class mAP50-95 from the validation-only comparison:

YOLO26m is slightly stronger on person, helmet, and gloves. YOLO26s is slightly stronger on vest and clearly stronger on goggles. The overall performance difference between YOLO26s and YOLO26m is extremely small, which supports selecting YOLO26s because it retains nearly identical detection quality with substantially lower inference cost.

## Final Selection

| Decision | Model |
|---|---|
| Absolute mAP50-95 winner | YOLO26m |
| Selected final model | YOLO26s |

YOLO26m has the highest validation mAP50-95, but the final validation-only comparison showed an advantage of only 0.00023 over YOLO26s. YOLO26s is selected because it provides nearly identical validation quality with substantially lower model size and compute cost.
