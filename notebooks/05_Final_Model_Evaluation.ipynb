# Stage 5: Final Model Evaluation

## 1. Final Model

The final selected model is **YOLO26s**, stored as `models/best.pt`.

YOLO26s was selected during the validation-only model comparison stage before opening the test split. The selection favored YOLO26s because it delivered nearly identical validation quality to YOLO26m while being faster and lighter for practical deployment.


## 2. Test Evaluation Protocol

| Item | Value |
|---|---:|
| Model | YOLO26s |
| Checkpoint | `models/best.pt` |
| Dataset YAML | `data/final_dataset/data.yaml` |
| Evaluation split | test |
| Test images | 323 |
| Test annotations | 2,523 |
| Image size | 640 |

The test split was not used for model selection. It was used only once as the final held-out evaluation set after the final model had already been chosen from validation results.


## 3. Overall Test Performance

| Metric | Test Result |
|---|---:|
| Precision | 0.8846 |
| Recall | 0.7708 |
| mAP50 | 0.8371 |
| mAP50-95 | 0.5833 |

The final held-out test performance is very close to the validation performance observed during model comparison, which supports that the selected model generalizes consistently across the project splits.


## 4. Per-Class Performance

| Class | Precision | Recall | mAP50 | mAP50-95 |
|---|---:|---:|---:|---:|
| person | 0.943 | 0.904 | 0.931 | 0.710 |
| helmet | 0.943 | 0.927 | 0.960 | 0.782 |
| vest | 0.908 | 0.819 | 0.885 | 0.604 |
| gloves | 0.792 | 0.490 | 0.610 | 0.354 |
| goggles | 0.837 | 0.714 | 0.799 | 0.466 |

Helmet and person are the strongest classes. Vest performance is also solid, though below helmet and person. Gloves are the weakest class, especially recall, which indicates that missed glove detections are the main remaining model limitation. Goggles perform moderately, with stronger precision than recall.

The measured inference speed confirms the practical advantage of selecting YOLO26s over the slightly higher-scoring YOLO26m. The model keeps strong detection quality while remaining lightweight enough for faster safety-inspection workflows.


## 8. Strengths and Limitations

**Strengths**

| Area | Evidence |
|---|---|
| Helmet detection | Highest class mAP50-95 at 0.782 and recall at 0.927 |
| Person detection | Strong precision, recall, and mAP50-95 |
| Vest detection | Solid overall performance with mAP50-95 of 0.604 |
| Runtime | 4.6 ms/image inference speed |

**Limitations**

| Area | Evidence |
|---|---|
| Gloves | Lowest recall at 0.490 and lowest mAP50-95 at 0.354 |
| Goggles | Moderate mAP50-95 at 0.466 and recall at 0.714 |
| Small PPE | Small objects remain harder to localize precisely under stricter IoU thresholds |

The final model is strongest on person and helmet detection and should be interpreted more carefully for gloves, where missed detections are the clearest remaining risk.


## 9. Final Evaluation Conclusion

YOLO26s achieved a final held-out test mAP50-95 of **0.5833**, with strong person and helmet performance and acceptable overall PPE detection behavior for the Smart Factory Safety Inspector project.

The test result is close to validation performance, which supports the model-selection decision made before test evaluation. The main limitation is glove detection recall, so downstream compliance decisions involving gloves should be treated with extra caution and reviewed during inference-stage testing.

**Final Stage 5 conclusion:** the selected model is validated on the held-out test split and is ready for the next project stage.
