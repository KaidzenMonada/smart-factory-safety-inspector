# Data Processing

This notebook documents the minimal confirmed dataset corrections applied after EDA manual review. It contains results only and no executable code.

## 1. Objective

Apply only the confirmed dataset fixes needed before training consideration:

- Remove three confirmed cross-split same-source leakage copies and their label files.
- Remove fourteen confirmed within-split same-source variants and their label files.
- Remove three false `goggles` annotations that visually represented person regions instead of goggles.
- Add seven missing `person` annotations for the four confirmed no-person label files.

No heuristic candidates, small objects, tiny boundary overflows, or multiple same-class PPE cases were automatically changed.
